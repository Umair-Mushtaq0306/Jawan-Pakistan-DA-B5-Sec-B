```python
open_file = open('student data.csv',mode='r')
open_file
```

```python
reader = csv.reader(open_file)
reader
```

```python
data = list(reader)
data
```

```python
header = data[0]
header
```

```python
data[1]
```

```python
student_data = data[1:]
student_data
```

```python
for phy in student_data:
    print(phy[3])
```

```python
for math in student_data:
    print(math[4])
```

```python
for math in student_data:
    print(type(int(math[4])))
```

```python
for per in student_data:
    chem = int(per[2])
    phy = int(per[3])
    math = int(per[4])
    percentage = round((chem + phy + math)/300*100,2)
    print(percentage)
    per.append(percentage)
```
