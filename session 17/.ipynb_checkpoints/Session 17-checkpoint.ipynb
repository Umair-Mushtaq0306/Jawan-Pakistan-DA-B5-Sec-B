```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
    
```

# Distinct

```python
q = 'select distinct(jobtitle) from salaries'
sql_query(q)
```

```python
q = 'select distinct jobtitle from salaries'
sql_query(q)
```

```python
q = '''select distinct(jobtitle) from salaries
        where jobtitle != 'Not provided'
'''
df = sql_query(q)
df
```

```python
q = """select employeename from salaries
        where employeename == 'Not provided'
        or
        employeename == 'Not Provided'
        or 
        employeename == 'not provided'


"""
sql_query(q)
```

```python
q = """select * from salaries

"""
sql_query(q)
```

```python
q = """select employeename,totalpay from salaries
        where employeename != 'Not provided'
        and
        employeename != 'Not Provided'
        and
        totalpay > 50000
       


"""
sql_query(q)
```

```python
# desc
# asc
# limit 20 

q = """select totalpay from salaries order by totalpay

"""
sql_query(q)
```
