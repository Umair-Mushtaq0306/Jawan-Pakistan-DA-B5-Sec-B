```python
# Payment Processing System

class Payment:
    def process_payment(self, amount):
        print(f"Processing payment of ${amount}.")
        return amount

class CreditCardPayment(Payment):
    def process_payment(self, amount):
        super().process_payment(amount)
        print(f"  - Credit card fee: ${amount * 0.03:.2f}")
        print(f"  - Total charged: ${amount * 1.03:.2f}")
        return amount * 1.03

class PayPalPayment(Payment):
    def process_payment(self, amount):
        super().process_payment(amount)
        print(f"  - PayPal fee: ${amount * 0.025:.2f}")
        print(f"  - Total charged: ${amount * 1.025:.2f}")
        return amount * 1.025

class CashPayment(Payment):
    def process_payment(self, amount):
        super().process_payment(amount)
        print(f"  - No fees for cash payment")
        return amount

# Testing
print("=" * 40)
print("CREDIT CARD PAYMENT")
print("=" * 40)
cc = CreditCardPayment()
cc.process_payment(100)

print("\n" + "=" * 40)
print("PAYPAL PAYMENT")
print("=" * 40)
pp = PayPalPayment()
pp.process_payment(100)

print("\n" + "=" * 40)
print("CASH PAYMENT")
print("=" * 40)
cash = CashPayment()
cash.process_payment(100)
```

```python
class Person:
    def __init__(self, name, age):
        self._name = name
        self._age = age
    
    # Getter for name
    def get_name(self):
        return self._name
    
    # Setter for name with validation
    def set_name(self, new_name):
        if new_name and len(new_name) > 0:
            self._name = new_name
        else:
            print("Error: Name cannot be empty")
    
    # Getter for age
    def get_age(self):
        return self._age
    
    # Setter for age with validation
    def set_age(self, new_age):
        if isinstance(new_age, int) and new_age > 0:
            self._age = new_age
        else:
            print("Error: Age must be a positive integer")
    
    def display_info(self):
        print(f"Name: {self._name}, Age: {self._age}")
    
    def __str__(self):
        return f"Person(name={self._name}, age={self._age})"


class Student(Person):
    def __init__(self, name, age, student_id=None, grade=None):
        super().__init__(name, age)
        self._student_id = student_id
        self._grade = grade
        self._courses = []
    
    # Additional getters and setters for student
    def get_student_id(self):
        return self._student_id
    
    def set_student_id(self, student_id):
        self._student_id = student_id
    
    def get_grade(self):
        return self._grade
    
    def set_grade(self, grade):
        self._grade = grade
    
    def add_course(self, course):
        self._courses.append(course)
        print(f"Added course: {course}")
    
    def get_courses(self):
        return self._courses
    
    def display_info(self):
        # Override parent method
        super().display_info()
        print(f"Student ID: {self._student_id}")
        print(f"Grade: {self._grade}")
        print(f"Courses: {', '.join(self._courses) if self._courses else 'No courses enrolled'}")
    
    def __str__(self):
        return f"Student(name={self._name}, age={self._age}, id={self._student_id})"


# Testing the code
print("=" * 40)
print("PERSON DEMONSTRATION")
print("=" * 40)

# Create a person
person1 = Person("John", 25)
person1.display_info()

# Accessing private attribute (works but not recommended)
print(f"\nDirect access to _name: {person1._name}")
print(f"Using getter: {person1.get_name()}")

# Using setters
person1.set_name("John Doe")
person1.set_age(26)
person1.display_info()

print("\n" + "=" * 40)
print("STUDENT DEMONSTRATION")
print("=" * 40)

# Create a student
student1 = Student("sara", 33, "S12345", "A")
print(student1)  # Uses __str__

print("\nStudent Info:")
student1.display_info()

# Add courses
student1.add_course("Mathematics")
student1.add_course("Physics")
student1.add_course("Computer Science")

print("\nUpdated Student Info:")
student1.display_info()

# Update student details
student1.set_name("Sara Johnson")
student1.set_age(34)
student1.set_grade("A+")

print("\nAfter Updates:")
student1.display_info()

# Accessing inherited methods
print(f"\nName (from getter): {student1.get_name()}")
print(f"Age (from getter): {student1.get_age()}")
```

```python
class Calculator:
    def __init__(self):
        self.result = 0
    
    def add(self, a, b):
        return a + b
    
    def subtract(self, a, b):
        return a - b
    
    def multiply(self, a, b):
        return a * b
    
    def divide(self, a, b):
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"


# Test the calculator
calc = Calculator()

print(f"10 + 5 = {calc.add(10, 5)}")
print(f"10 - 5 = {calc.subtract(10, 5)}")
print(f"10 * 5 = {calc.multiply(10, 5)}")
print(f"10 / 5 = {calc.divide(10, 5)}")
print(f"10 / 0 = {calc.divide(10, 0)}")
```

```python
class BankAccount:
    def __init__(self, account_holder, initial_balance=0):
        self.account_holder = account_holder
        self.balance = initial_balance
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited: ${amount:.2f}")
            print(f"New balance: ${self.balance:.2f}")
        else:
            print("Deposit amount must be positive")
    
    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                print(f"Withdrawn: ${amount:.2f}")
                print(f"New balance: ${self.balance:.2f}")
            else:
                print(f"Insufficient balance! Available: ${self.balance:.2f}")
        else:
            print("Withdrawal amount must be positive")
    
    def check_balance(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: ${self.balance:.2f}")


# Create an account
account = BankAccount("John Doe", 1000)

# Perform operations
account.check_balance()
account.deposit(500)
account.withdraw(200)
account.withdraw(1500)  # Should fail (insufficient balance)
account.check_balance()
```

```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary
        self.department = "General"
        self.projects = []

    def work(self):
        print(f'{self.name} is working on {len(self.projects)} project(s)')

    def get_details(self):
        print(f'Name: {self.name}, Employee ID: {self.employee_id}, Salary: ${self.salary:,}')
        print(f'Department: {self.department}')

    def assign_project(self, project_name):
        self.projects.append(project_name)
        print(f'{self.name} assigned to project: {project_name}')

    def get_salary_info(self):
        monthly = self.salary / 12
        weekly = self.salary / 52
        print(f'Annual: ${self.salary:,}')
        print(f'Monthly: ${monthly:,.2f}')
        print(f'Weekly: ${weekly:,.2f}')


class Manager(Employee):
    def __init__(self, name, employee_id, salary, team_size, department="Management"):
        super().__init__(name, employee_id, salary)
        self.team_size = team_size
        self.department = department
        self.team_members = []
        self.bonus_percentage = 15

    def work(self):
        print(f'{self.name} is managing a team of {self.team_size} employees')
        if self.team_members:
            print(f'Team members: {", ".join(self.team_members)}')

    def get_details(self):
        super().get_details()
        print(f'Team Size: {self.team_size}')
        print(f'Bonus Percentage: {self.bonus_percentage}%')
        if self.team_members:
            print(f'Team Members: {", ".join(self.team_members)}')

    def add_team_member(self, employee_name):
        self.team_members.append(employee_name)
        self.team_size = len(self.team_members)
        print(f'{employee_name} added to team. Team size is now {self.team_size}')

    def calculate_bonus(self):
        bonus = self.salary * (self.bonus_percentage / 100)
        print(f'{self.name}\'s bonus: ${bonus:,.2f}')
        return bonus

    def set_bonus_percentage(self, percentage):
        if 0 <= percentage <= 100:
            self.bonus_percentage = percentage
            print(f'Bonus percentage updated to {percentage}%')
        else:
            print('Bonus percentage must be between 0 and 100')


class Developer(Employee):
    def __init__(self, name, employee_id, salary, programming_languages=None):
        super().__init__(name, employee_id, salary)
        self.programming_languages = programming_languages or []
        self.department = "Engineering"
        self.lines_of_code = 0

    def work(self):
        print(f'{self.name} is writing code in {", ".join(self.programming_languages)}')
        print(f'Lines of code written today: {self.lines_of_code}')

    def get_details(self):
        super().get_details()
        print(f'Programming Languages: {", ".join(self.programming_languages)}')
        print(f'Total Lines of Code: {self.lines_of_code}')

    def add_language(self, language):
        if language not in self.programming_languages:
            self.programming_languages.append(language)
            print(f'{language} added to skills')
        else:
            print(f'{language} already in skills')

    def code(self, lines):
        self.lines_of_code += lines
        print(f'Coded {lines} lines. Total: {self.lines_of_code} lines')


class Salesperson(Employee):
    def __init__(self, name, employee_id, salary, sales_target=100000):
        super().__init__(name, employee_id, salary)
        self.sales_target = sales_target
        self.sales_made = 0
        self.department = "Sales"
        self.commission_rate = 0.05

    def work(self):
        progress = (self.sales_made / self.sales_target) * 100 if self.sales_target > 0 else 0
        print(f'{self.name} is selling products')
        print(f'Sales target progress: {progress:.1f}%')

    def get_details(self):
        super().get_details()
        print(f'Sales Target: ${self.sales_target:,}')
        print(f'Sales Made: ${self.sales_made:,}')
        print(f'Commission Rate: {self.commission_rate * 100}%')
        print(f'Commission Earned: ${self.calculate_commission():,.2f}')

    def make_sale(self, amount):
        self.sales_made += amount
        commission = amount * self.commission_rate
        print(f'Sale of ${amount:,} made! Commission earned: ${commission:,.2f}')

    def calculate_commission(self):
        return self.sales_made * self.commission_rate

    def set_commission_rate(self, rate):
        if 0 <= rate <= 0.20:
            self.commission_rate = rate
            print(f'Commission rate updated to {rate * 100}%')
        else:
            print('Commission rate must be between 0 and 20%')


# Demonstration
print("=" * 50)
print("MANAGER EXAMPLE")
print("=" * 50)

# Create a manager
manager = Manager("Sarah Johnson", "M001", 95000, 0, "IT Operations")
manager.add_team_member("John Doe")
manager.add_team_member("Jane Smith")
manager.add_team_member("Bob Wilson")
manager.work()
print("\nManager Details:")
manager.get_details()
manager.calculate_bonus()

print("\n" + "=" * 50)
print("DEVELOPER EXAMPLE")
print("=" * 50)

# Create a developer
dev = Developer("Alice Chen", "D001", 75000, ["Python", "JavaScript", "React"])
dev.assign_project("E-commerce Website")
dev.assign_project("Mobile App")
dev.code(150)
dev.code(200)
dev.work()
print("\nDeveloper Details:")
dev.get_details()
dev.add_language("TypeScript")

print("\n" + "=" * 50)
print("SALESPERSON EXAMPLE")
print("=" * 50)

# Create a salesperson
sales = Salesperson("Mike Robinson", "S001", 65000, 250000)
sales.make_sale(45000)
sales.make_sale(32000)
sales.make_sale(28000)
sales.work()
print("\nSalesperson Details:")
sales.get_details()
sales.set_commission_rate(0.07)
sales.calculate_commission()
```
