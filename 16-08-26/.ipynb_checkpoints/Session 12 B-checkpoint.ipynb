## ***12. Find Minimum, Maximum, Average BasepPay***

```python
df['BasePay'] = pd.to_numeric(df['BasePay'], errors='coerce')

# Calculate statistics
min_base_pay = df['BasePay'].min()
max_base_pay = df['BasePay'].max()
avg_base_pay = df['BasePay'].mean()

print(f"Minimum BasePay: {min_base_pay}")
print(f"Maximum BasePay: {max_base_pay}")
print(f"Average BasePay: {avg_base_pay:.2f}")
```

```python
df.BasePay = df.BasePay.replace('Not Provided',0).astype(float)
df.BasePay
```

```python
##22. Find the Number of Employees per Year
```

```python
# 23. Find the Average Overtime Pay by Year
```

```python
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')
avg_overtime_per_year = df.groupby('Year')['OvertimePay'].mean()
print(avg_overtime_per_year)
```

```python
# 24. Find the Total Payroll for Each Year
```

```python
total_payroll_per_year = df.groupby('Year')['TotalPay'].sum()
print(total_payroll_per_year)
```

```python
# 25. Find the Most Common First Names
```

```python
df['FirstName'] = df['EmployeeName'].str.split().str[0]
most_common_names = df['FirstName'].value_counts().head(10)
print(most_common_names)
```

```python
# 26. Find Employees Who Earned $0 Total Pay
```

```python
zero_pay_employees = df[df['TotalPay'] == 0]
print(f"Number of employees with $0 TotalPay: {len(zero_pay_employees)}")
print(zero_pay_employees[['EmployeeName', 'JobTitle', 'Year']].head(10))
```

```python
# 27. Find the Average Total Pay by Job Title (Top 10)
```

```python
avg_pay_by_job = df.groupby('JobTitle')['TotalPay'].mean().sort_values(ascending=False).head(10)
print(avg_pay_by_job)
```

```python
# 28. Find the Year with the Highest Average Pay
```

```python
highest_avg_pay_year = df.groupby('Year')['TotalPay'].mean().idxmax()
highest_avg_pay_value = df.groupby('Year')['TotalPay'].mean().max()
print(f"Year with highest average pay: {highest_avg_pay_year} - ${highest_avg_pay_value:.2f}")
```

```python
# 29. Count Employees by Status
```

```python
status_counts = df['Status'].value_counts(dropna=False)
print(status_counts)
```

```python
# 30. Calculate the Correlation Between BasePay and TotalPay
```

```python
correlation = df['BasePay'].corr(df['TotalPay'])
print(f"Correlation between BasePay and TotalPay: {correlation:.4f}")
```
