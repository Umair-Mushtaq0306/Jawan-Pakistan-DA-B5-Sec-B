# Logical Operators in Pandas:
* & (and)
* | (or)
* ~ (not)    tilde

```python
df
```

```python
df['Region']
```

```python
df['Region'].unique()
```

```python
df['Region'] == 'South'
```

```python
df[df['Region'] == 'South']
```

```python
df[df['Sales'] > 200]
```

```python
df[(df['Region'] == 'South') & (df['Sales'] > 200)]
```

```python
df[(df['Region'] == "South")|(df['Region'] == "West")]
```

```python
df[~(df['Region'] == "South")]
```
