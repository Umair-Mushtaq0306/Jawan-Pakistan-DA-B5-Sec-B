```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('cars.csv')
df.head()
```

```python
df.tail()
```

```python
#dropping irrelivent columns
```

```python
df['Vehicle Size'].unique()
```

```python
df.shape
```

```python
df = df.drop(['Vehicle Size'],axis=1)
df
```

```python
#Rename
```

```python
df = df.rename(columns={'Engine HP': 'HP','Engine Cylinders':'Cylinders','Transmission Type':'Transmission','Driven_Wheels':'Drive Mode','highway MPG':'MGP-H','city mpg':'MGP-C'})
df
```

```python
# Duplicate Rows
```

```python
df.shape
```

```python
df[df.duplicated()]
```

```python
dup_rows = df[df.duplicated()]
print('Number of Duplicated Rows: ', dup_rows.shape)
```

```python
df.count()
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
df.count()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
df.count()
```

```python
df.Make
```

```python
df.Make.value_counts()
```

```python
df.Make.value_counts().plot(kind='bar')
```

```python
df.Make.value_counts().plot(kind='pie')
```

```python
df.Make.value_counts().plot(kind='bar')
plt.title('Number of cars by make')
plt.xlabel('Make')
plt.ylabel('Number of Cars')
plt.show()
```

```python
df.Make.value_counts().head(10).plot(kind='bar')
plt.title('Number of cars by make')
plt.xlabel('Make')
plt.ylabel('Number of Cars')
plt.show()
```
