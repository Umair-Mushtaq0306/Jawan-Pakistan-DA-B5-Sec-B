```python
def financial_report(company_name, revenue, expenses, revenue_target):
  profit = revenue - expenses

  if profit != 0:
    profit_margin = round(((profit/revenue)*100),2)
  else:
    profit_margin = 0

  revenue_variance = revenue - revenue_target

  print("_____________FINANCIAL_REPORT________________\n")
  print("Company Name: ", company_name)
  print("Revenue in PKR: ", revenue)
  print("Expenses in PKR: ", expenses)
  print("Profit in PKR: ", profit)
  print("Profit Margin: ", profit_margin,"%")
  print("Variance in PKR: ", revenue_variance)

  if revenue >= revenue_target:
    print("We have acheived our target!")
  else:
    print("Target has not been acheived!")

  if profit_margin > 20:
    print("Business is doing great!")
  elif profit_margin > 13:
    print("Business is doing good, but we can improve")
  else:
    print("Need improvements in ourr business!")
```
