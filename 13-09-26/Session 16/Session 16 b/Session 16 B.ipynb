# Load Multiple Sheets from Excel

```python
import pandas as pd
```

```python
pd.read_excel('Sales_Data.xlsx')
```

```python
all_sheets = pd.read_excel('Sales_Data.xlsx', sheet_name=None)
all_sheets
```

```python
all_sheets.keys()
```

```python
all_sheets.items()
```

```python
for name, df in all_sheets.items():
    print(name, df.shape)
```

```python
combined_df = pd.concat([df.assign(Month=name) for name, df in all_sheets.items()], ignore_index=True)
combined_df
```

```python
combined_df.info()
```

```python
combined_df['Total_Sales'] = combined_df['Quantity'] * combined_df['Price']
combined_df.head()
```

```python
combined_df.groupby('Month')['Total_Sales'].sum()
```

```python
combined_df.to_csv('Combined_Sales_Data.csv', index=False)
```
