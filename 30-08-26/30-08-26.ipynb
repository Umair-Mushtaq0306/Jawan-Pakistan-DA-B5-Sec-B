```python
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
```

```python
print(Q1)
print(Q3)
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 - 1.5 * IQR
print(lower_bound)
print(upper_bound)
```

```python
df_filter = df[df['Age']>= lower_bound]
```

```python
print(df[df.index.isin(df_filtered.index)])

```

```python
# Create boxplot for Age by Passenger Class
plt.figure(figsize=(10, 7))
sns.boxplot(x='Pclass', y='Age', data=df, palette='GnBu_d')
plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()
```
